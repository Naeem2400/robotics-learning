# Lesson 45 — First SLAM Robot Simulation

No hardware. A MacBook runs Webots; Webots runs a TurtleBot3; the TurtleBot3 carries a virtual LiDAR.

This notebook is the **maths**. The robot lives in `worlds/slam_lab.wbt` — open that in Webots and press Run after you have walked through the cells.

## 1. World vs robot vs controller

| Piece | File | Job |
|---|---|---|
| World | `worlds/slam_lab.wbt` | Floor, walls, the hallway |
| Robot | TurtleBot3 Burger inside that world | Motors + `LDS-01` lidar |
| Controller | `controllers/slam_scout/slam_scout.py` | Python that reads the lidar and drives |

Webots is a virtual robotics laboratory. The architecture is the same as a real robot: sensor → decision → motor. Only the sensor is simulated.

## 2. A lidar scan is a list of metres

360 beams. Middle of the list is forward. Bigger number = further away — the **opposite** of the e-puck IR sensors in Lesson 26.

````python
from lidar_scan import fake_corridor, front_left_right, should_stop, describe

scan = fake_corridor(front=2.4, left=1.2, right=0.8)
print(f"{len(scan)} beams")
print(describe(scan))
````

## 3. The wall approaches

Same hallway, the front range shrinking. This is what you will watch in the Webots console.

````python
for metres in (5.0, 4.0, 3.0, 2.0, 1.0, 0.7):
    front, left, right = front_left_right(fake_corridor(front=metres))
    action = "STOP" if should_stop(front) else "FORWARD"
    print(f"front {front:4.1f} m   left {left:4.1f} m   right {right:4.1f} m   →  {action}")
````

## 4. One scan is a snapshot, not a map

`R` is the robot. `#` is a lidar hit. This is **one pose**. SLAM is many of these, stitched with odometry (Lesson 44) and corrected by loop closure.

````python
from lidar_scan import occupancy_from_scan, render_grid

print(render_grid(occupancy_from_scan(fake_corridor(front=2.4, left=1.2, right=0.8))))
````
