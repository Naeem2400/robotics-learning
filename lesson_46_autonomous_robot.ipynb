````python
from lidar_scan import choose_action

print("open path     ", choose_action(front=2.4, left=1.2, right=0.8))
print("wall, left open", choose_action(front=0.5, left=2.0, right=0.8))
print("wall, right open", choose_action(front=0.5, left=0.8, right=2.0))
````

## Two wheels

| Left | Right | Motion |
|---|---|---|
| same + | same + | forward |
| − | + | spin left |
| + | − | spin right |
